## Jeu de données

```python
np.random.seed(123)

mu, sigma = 2, 5
X_train_1_1 = np.random.normal(mu, sigma, 100)

mu, sigma = 5, 1
X_train_1_2 = np.random.normal(mu, sigma, 100)

mu, sigma = 5, 3
X_train_2_1 = np.random.normal(mu, sigma, 100)

mu, sigma = 20, 1
X_train_2_2 = np.random.normal(mu, sigma, 100)

mu, sigma = 22, 22.5
X_train_3_1 = np.random.normal(mu, sigma, 100)

mu, sigma = 10, 1
X_train_3_2 = np.random.normal(mu, sigma, 100)
```

```python
X_train = np.stack([
        np.concatenate([X_train_1_1, X_train_2_1, X_train_3_1]),
        np.concatenate([X_train_1_2, X_train_2_2, X_train_3_2])], axis=1)

np.random.shuffle(X_train)
```

```python
plt.scatter(X_train[:, 0], X_train[:, 1])
plt.show()
```

```python
from hierarchical_clustering import HierarchicalClustering as HC
```

```python
agglo = HC(k=3)
agglo.fit(X_train)
```

```python
for i in range(len(agglo.clusters)):
    plt.scatter(agglo.clusters[i][0], agglo.clusters[i][1])
plt.show()
```
